```python
from sql_metadata import Parser

# A complex SQL statement containing CTE, subqueries, and multiple JOINs
sql_query = """
SELECT t1.document_name FROM documents AS t1 JOIN document_sections AS t2 ON t1.document_code  =  t2.document_code GROUP BY t1.document_code ORDER BY count(*) DESC LIMIT 1
"""

# Create parser instance
parser = Parser(sql_query)

# Directly access the .tables property to get all table names
table_names = parser.tables

print(f"Original SQL statement:\n{sql_query}")
print("="*50)
print(f"Parsed table names: {table_names}")

# You can also get more detailed information
# print(f"Column names: {parser.columns}")
# print(f"Table names with aliases: {parser.tables_aliases}")
```
